```python
new_predicted_outputs = np.array([x['new_predicted_output'][0] for x in campaign['optimizer']])
new_predicted_uncertainties = 2*np.sqrt(np.array([x['new_predicted_uncertainty'][0] for x in campaign['optimizer']]))
plt.plot(new_predicted_outputs[::-1])
plt.plot(np.array(runs['dep_rate_fit'])[::-1])
plt.fill_between(np.arange(len(new_predicted_outputs)), 
                 (new_predicted_outputs-new_predicted_uncertainties)[::-1],
                 (new_predicted_outputs+new_predicted_uncertainties)[::-1], color='r')
```
